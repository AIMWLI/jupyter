## match/case (3.10)

类似 switch 的模式匹配，写起来比 if/elif 清爽。

```python
def handle_status(code):
    match code:
        case 200:
            return "OK"
        case 400 | 401:
            return "Client error"
        case 500:
            return "Server error"
        case _:
            return "Unknown"

print(handle_status(200))
print(handle_status(404))
```

可以解包结构，处理数据很方便。

```python
def process_point(point):
    match point:
        case (0, 0):
            print("原点")
        case (x, 0):
            print(f"x 轴, x={x}")
        case (0, y):
            print(f"y 轴, y={y}")
        case (x, y):
            print(f"点 ({x}, {y})")
        case _:
            print("不是二维点")

process_point((0, 0))
process_point((3, 0))
process_point((1, 2))
```

## 异常链 + ExceptionGroup (3.11)

同时抛多个异常的场景终于有原生支持了。

```python
def validate(data):
    errors = []
    if "name" not in data:
        errors.append(ValueError("缺少 name"))
    if "age" not in data:
        errors.append(ValueError("缺少 age"))
    if data.get("age", 0) < 0:
        errors.append(ValueError("age 不能为负"))
    if errors:
        raise ExceptionGroup("校验失败", errors)
    return True

try:
    validate({"name": "test", "age": -1})
except* ValueError as e:
    print(f"ValueError: {e.exceptions}")
```

## Type Hint 相关

3.10+ 的类型标注越写越舒服了。

```python
from typing import TypeVar

# 3.12 可以用 type 别名
type Vector = list[float]

# TypeVar 可以加类型约束
T = TypeVar("T", int, float)

def add(a: T, b: T) -> T:
    return a + b

print(add(1, 2))
print(add(1.5, 2.5))
```
